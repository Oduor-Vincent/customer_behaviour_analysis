```python
# Filling missing Review Rating values with median rating of their respective Category
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming the columns into snake case format for easy documentation
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# Creating a new column called "generation"
bins = [14, 30, 46, 62, 81] # Defines age threshold
labels = ["Gen Z", "Millenials", "Gen X", "Boomers"] # Defines the labels for each age category

# Creating the generation column
# 'right=False' makes intervals inclusive of left edge and exclusive of right edge [14, 30) -- Gen Z is of the age 14 to 29 yrs
df["generation"] = pd.cut(df["age"], bins=bins, labels=labels, right=False)
```

```python
# Checking the column 'generation' created
df[["age","generation"]].sample(50)
```

```python
# Creating 'purchase_frequency_days' column
frequency_mapping = {
    "Fortnightly": 14,
    "Weekly": 7,
    "Bi-Weekly": 14,
    "Annually": 365,
    "Quarterly": 90,
    "Monthly": 30,
    "Every 3 Months": 90    
}

df["purchase_frequency_days"] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].sample(15)
```

```python
# Drop "promo_code_used" column -- a redundant column
df = df.drop("promo_code_used", axis=1)
```
